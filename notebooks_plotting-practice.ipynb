```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1
sns.boxplot(
    data=df,
    x="Department",
    y="Salary",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Salary by Department")

# Plot 2
sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Salary vs Age")

# Plot 3
sns.boxplot(
    data=df,
    x="Job_Level",
    y="Salary",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Salary by Job Level")

# Plot 4
sns.scatterplot(
    data=df,
    x="Experience_Years",
    y="Salary",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Salary vs Experience")

plt.tight_layout()
plt.show()
```
